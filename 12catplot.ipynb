# catplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var=sns.load_dataset("tips").head(20)
var
```

```python
sns.catplot(x="tip",y="size",data=var)
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=var,hue="sex",palette="plasma",height=5)
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=var,hue="sex",palette="plasma",kind="bar")
#kind=strip,swarm,box,violin,boxen,point,bar,count
plt.show()
```
