# Histogram

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
y1=sns.load_dataset("penguins")
y1
```

```python
sns.displot(y1["bill_length_mm"])
plt.show()
```

```python
sns.displot(y1["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240],
            kde=True,rug=True,color="g")
plt.show()
```

```python
sns.displot(y1["flipper_length_mm"],log_scale=True)
plt.show()
```
