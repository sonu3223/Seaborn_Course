# Vionlin

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
y1=sns.load_dataset("tips")
y1
```

```python
sns.violinplot(x="day",y="total_bill",data=y1)
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=y1,hue="time")
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=y1,linewidth=2,color="r") #palette="Dark2",saturation=0.7
plt.show()
```

```python
sns.violinplot(x="time",y="total_bill",data=y1,order=["Dinner","Lunch"])
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=y1,hue="sex",split=True,scale="count") #scale="count",area,width
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=y1,hue="sex")
plt.show()
```

```python
sns.violinplot(y="day",x="total_bill",data=y1,hue="sex")
plt.show()
```

```python
sns.violinplot(x="time",y="total_bill",data=y1,inner="quart") #inner="quart",box,quartlite,point,stick
plt.show()
```

```python
sns.violinplot(y=y1["total_bill"])
plt.show()
```
