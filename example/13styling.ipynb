```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var=sns.load_dataset("tips")
var
```

```python
sns.barplot(x="day",y="total_bill",data=var)
plt.grid()
plt.show()
```

```python
sns.set_style("darkgrid") #darkgrid,ticks ,whitegrid
sns.boxplot(y=var["total_bill"])
# plt.grid()
plt.show()
```

```python
sns.set_style("whitegrid")
sns.barplot(x="day",y="total_bill",data=var)
plt.figure(figsize=(3,12))
# plt.grid()
sns.despine()
plt.show()
```

```python

sns.set_style("whitegrid")
sns.set_context("poster",font_scale=.6) #poster=paper
sns.barplot(x="day",y="total_bill",data=var,palette="cool")
plt.show()
```
