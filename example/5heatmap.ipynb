# Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
var=np.linspace(1,10,20).reshape(5,4)
var
```

```python

sns.heatmap(var,vmin=0,vmax=10,cmap="PuOr",annot=True)
plt.show()
```
