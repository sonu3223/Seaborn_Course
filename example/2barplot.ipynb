# Barplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
y1=sns.load_dataset("penguins")
y1
```

```python
# sns.barplot(x=y1.island,y=y1.bill_length_mm)
sns.barplot(x="island",y="bill_length_mm",data=y1)
plt.show()
```

```python

sns.barplot(x="island",y="bill_length_mm",data=y1,hue="sex")
plt.show()
```

```python
order1=["Biscoe","Dream","Torgersen"]
sns.barplot(x="island",y="bill_length_mm",data=y1,hue="sex",order=order1)
plt.show()
```

```python
order1=["Biscoe","Dream","Torgersen"]
sns.barplot(x="island",y="bill_length_mm",data=y1,hue="sex",order=order1,
            hue_order=["Female","Male"],ci=100,n_boot=2) #,dodge=False
plt.show()
```

```python
order1=["Biscoe","Dream","Torgersen"]
sns.barplot(x="flipper_length_mm",y="bill_length_mm",data=y1,hue="sex",orient="h")
plt.show()
```

```python
sns.barplot(x="island",y="bill_length_mm",data=y1,color="b",palette="Accent")
plt.show()
```

```python
# sns.set(style="darkgrid")
sns.barplot(x="island",y="bill_length_mm",data=y1,saturation=200,errcolor="r",errwidth=2,
           capsize=.1,alpha=.5)
plt.show()
```
