# Scatter plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
y1=sns.load_dataset("penguins") #.head(20)
y1
```

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=y1)
plt.show()
```

```python
n={"Male":"*","Female":"o"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=y1,hue="sex",style="sex",
                size="sex",sizes=(100,200),palette="Accent",alpha=.9,markers=n)
plt.show()
```
