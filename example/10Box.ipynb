# box

```python
 import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var=sns.load_dataset("tips")
var
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var)
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",palette="plasma",order=["Fri","Sun","Thu","Sat"],showmeans=True,
            meanprops={"marker":"+","markeredgecolor":"black"},linewidth=2) #color="r"
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(y="day",x="total_bill",data=var,hue="sex",palette="plasma",order=["Fri","Sun","Thu","Sat"],showmeans=True,
            meanprops={"marker":"+","markeredgecolor":"black"},linewidth=2) #color="r"
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(data=var,palette="plasma",orient="h") #color="r"
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(y=var["total_bill"]) #color="r"
plt.show()
```
