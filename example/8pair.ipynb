# Pairplot

```python
 import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
y1=sns.load_dataset("tips").head(20)
y1
```

```python
sns.pairplot(y1)
plt.show()
```

```python
sns.pairplot(y1,hue="sex")
plt.show()
```

```python
sns.pairplot(y1,vars=["total_bill","tip"],hue="sex",hue_order=["Female","Male"]) #palette="BuGn"
plt.show()
```

```python
sns.pairplot(y1,hue="sex",hue_order=["Female","Male"],y_vars=["total_bill","tip"]) #palette="BuGn"
plt.show()
```

```python
sns.pairplot(y1,hue="sex",hue_order=["Female","Male"],kind="reg",diag_kind="hist") #kind="reg,scatter,kde,hist
plt.show()
```

```python
sns.pairplot(y1,hue="sex",hue_order=["Female","Male"],markers=["*",">"])
plt.show()
```
