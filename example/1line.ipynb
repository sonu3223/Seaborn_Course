# Line plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
var=[1,2,3,4,5,6]
var1=[5,4,2,1,7,4]
x1=pd.DataFrame({"var":var,"var1":var1})
sns.lineplot(x="var",y="var1",data=x1) 
# not ploted because this work on dataset jo ki pandas se fir seaborn
plt.show()
```

```python
plt.plot(var,var1) # matplotlib plot
plt.show()
```

```python
var=[1,2,3,4,5,6]
var1=[5,4,2,1,7,4]
sns.lineplot(x=var,y=var1)
plt.show()
```
