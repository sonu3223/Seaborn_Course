# Countplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
y1=sns.load_dataset("tips")
y1
```

```python
sns.countplot(x="sex",data=y1) # counting batata hai ye plot
plt.show()
```

```python
sns.barplot(x="sex",y="size",data=y1) #but this give the mean value
plt.show()
```

```python
sns.countplot(x="sex",data=y1,hue="smoker",color="b",saturation=.2) #y="sex" palette="RdGy"
plt.show()
```
