# Strip plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var=sns.load_dataset("tips")
var
```

```python
sns.stripplot(var)
plt.show()
```

```python
# b={"Male":"o","Female":"<"}
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",linewidth=1,edgecolor="g",size=5,marker="*",alpha=.2)
#jitter=1
plt.show()
```

```python

sns.stripplot(y=var["total_bill"])
plt.show()
```
